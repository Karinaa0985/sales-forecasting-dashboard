```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel("Superstore.xls")
monthly = df.set_index("Order Date").resample("MS")["Sales"].sum()
```

```python
print(monthly.groupby(monthly.index.year).sum().round())
```

```python
by_month = monthly.groupby(monthly.index.month).mean().round()
print(by_month)
by_month.plot(kind="bar", figsize=(10, 4), title="Average sales by calendar month")
plt.show()
```

```python
from statsmodels.tsa.holtwinters import ExponentialSmoothing

train, test = monthly.iloc[:-6], monthly.iloc[-6:]

def mape(actual, pred):
    return np.mean(np.abs((actual.values - pred.values) / actual.values)) * 100

def rmse(actual, pred):
    return np.sqrt(np.mean((actual.values - pred.values) ** 2))

results = {}

# Baseline 1: average of the last 12 training months
results["Moving average (12m)"] = pd.Series(train.iloc[-12:].mean(), index=test.index)

# Baseline 2: "same months last year"
results["Seasonal naive"] = pd.Series(train.iloc[-12:-6].values, index=test.index)

# Holt-Winters: trend + seasonality
m1 = ExponentialSmoothing(train, trend="add", seasonal="add", seasonal_periods=12).fit()
results["Holt-Winters (trend+season)"] = pd.Series(m1.forecast(6).values, index=test.index)

# Holt-Winters: seasonality only
m2 = ExponentialSmoothing(train, trend=None, seasonal="add", seasonal_periods=12).fit()
results["Holt-Winters (season only)"] = pd.Series(m2.forecast(6).values, index=test.index)

table = pd.DataFrame({
    name: {"MAPE %": mape(test, p), "RMSE": rmse(test, p)}
    for name, p in results.items()
}).T.round(2).sort_values("MAPE %")
print(table)
```

```python
plt.figure(figsize=(12, 5))
plt.plot(train.index, train, label="Train")
plt.plot(test.index, test, label="Actual (test)", color="black", linewidth=2)
for name, p in results.items():
    plt.plot(p.index, p, "--", label=name)
plt.legend()
plt.title("Forecast comparison on last 6 months")
plt.show()
```

```python
from statsmodels.tsa.holtwinters import ExponentialSmoothing

future_idx = pd.date_range(monthly.index[-1] + pd.offsets.MonthBegin(1),
                           periods=12, freq="MS")

# Benchmark: flat 12-month average
fc_ma = pd.Series(monthly.iloc[-12:].mean(), index=future_idx)

# Main model: Holt-Winters, seasonality only, refit on ALL 48 months
hw = ExponentialSmoothing(monthly, trend=None, seasonal="add",
                          seasonal_periods=12).fit()
fc_hw = pd.Series(hw.forecast(12).values, index=future_idx)

# Approximate 95% range from the model's typical error
sd = hw.resid.std()
lower = (fc_hw - 1.96 * sd).clip(lower=0)
upper = fc_hw + 1.96 * sd

print(fc_hw.round())
print("Total forecast 2013:", round(fc_hw.sum()))
print("Total actual 2012:", round(monthly.iloc[-12:].sum()))
```

```python
plt.figure(figsize=(12, 5))
plt.plot(monthly.index, monthly, label="Actual", color="black")
plt.plot(fc_hw.index, fc_hw, "--", label="Forecast (Holt-Winters)", color="tab:red")
plt.plot(fc_ma.index, fc_ma, ":", label="Benchmark (12m average)", color="tab:green")
plt.fill_between(fc_hw.index, lower, upper, alpha=0.2, color="tab:red", label="~95% range")
plt.legend()
plt.title("Sales forecast for 2013")
plt.show()
```

```python
actual = pd.DataFrame({"Date": monthly.index, "Actual": monthly.values})
fcast = pd.DataFrame({
    "Date": future_idx,
    "Forecast": fc_hw.values,
    "Lower": lower.values,
    "Upper": upper.values,
    "Benchmark": fc_ma.values,
})
out = pd.concat([actual, fcast], ignore_index=True)
out.to_csv("forecast_output.csv", index=False)

accuracy = pd.DataFrame({
    "Model": ["Moving average (12m)", "Holt-Winters (season only)",
              "Seasonal naive", "Holt-Winters (trend+season)"],
    "MAPE": [13.07, 14.59, 17.51, 19.40],
    "RMSE": [43482.79, 46650.32, 57507.93, 61991.08],
})
accuracy.to_csv("model_accuracy.csv", index=False)

print(out.tail(14))
```
